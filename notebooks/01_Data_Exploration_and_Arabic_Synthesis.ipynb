### 1.2 اختبار النواة البرمجية المكتوبة بلغة C++ (Fast LaTeX Syntax Guard)
- **الهدف:** اختبار وحدة C++ المترجمة (`libmath_core.so`) للتحقق الفوري من توازن الأقواس `{ } ( ) [ ]` والأكواد الرياضية في أجزاء من الميكروثانية.

```python
from khwarizmi.vision.syntax_guard import is_valid_latex_syntax, normalize_latex, compute_formula_hash

test_cases = [
    (r"\frac{-b \pm \sqrt{b^2 - 4ac}}{2a}", True),
    (r"\int_{0}^{\infty} x^2 e^{-x} dx", True),
    (r"\frac{a + b}{c", False),  # قوس مفقود
    (r"x^2 + (y - 1] = 0", False) # عدم تطابق الأقواس
]

print("=== فحص النواة الفائقة C++ للتحقق النحوي المسبق ===")
for latex, expected in test_cases:
    valid = is_valid_latex_syntax(latex)
    status = "✓ صالحة" if valid else "✗ غير صالحة (مرفوضة)"
    print(f"الصيغة: {latex:<40} | النتيجة: {status}")

sample = r"\frac{1}{2} * x + 4 = 10"
print(f"\nالصيغة المعيارية الموحدة: {normalize_latex(sample)}")
print(f"بصمة الهاش الفريدة (FNV-1a 64-bit): {hex(compute_formula_hash(sample))}")
```

### 1.3 المحرك الرمزي الصارم (SymPy Neuro-Symbolic Verification Engine)
- **الهدف:** تنفيذ الفصل العصبي-الرمزي؛ المحرك يحل المعادلات خطوة بخطوة ويفحص تكافؤ الخطوات جبرياً لمنع الهلوسة.

```python
from khwarizmi.symbolic.engine import solve_step_by_step, verify_step_equivalence

# 1. حل معادلة خطية خطوة بخطوة
print("--- 1. حل معادلة خطية (2x + 4 = 10) ---")
steps_linear = solve_step_by_step("2*x + 4 = 10")
for s in steps_linear:
    print(f"الخطوة {s['step_num']}: [{s['rule_name_ar']}] {s['latex']} | {s['explanation_ar']}")

# 2. حل معادلة تربيعية خطوة بخطوة
print("\n--- 2. حل معادلة تربيعية (x^2 - 5x + 6 = 0) ---")
steps_quad = solve_step_by_step("x**2 - 5*x + 6 = 0")
for s in steps_quad:
    print(f"الخطوة {s['step_num']}: [{s['rule_name_ar']}] {s['latex']}")
    print(f"   الشرح التربوي: {s['explanation_ar']}")

# 3. التحقق الرمزي من صحة خطوة الطالب (Neuro-Symbolic Gate)
print("\n--- 3. بوابة الفحص الرمزي للخطوات ---")
print("هل الانتقال من (2x + 4 = 10) إلى (2x = 6) صحيح؟", verify_step_equivalence("2*x + 4 = 10", "2*x = 6"))
print("هل الانتقال من (2x + 4 = 10) إلى (2x = 7) صحيح؟", verify_step_equivalence("2*x + 4 = 10", "2*x = 7"))
```

### 1.4 المولد الرمزي الهجين للحقيبة التعليمية العربية (Hybrid Arabic Math Synthesis)
- **الهدف:** توليد 100 ألف مسألة مشروحة باللغة العربية الفصحى مع التزام كامل بالبنية المهيكلة `[RULE]` و `[LATEX]` و `[EXPLANATION_AR]`.

```python
from khwarizmi.reasoning.generator import HybridArabicMathGenerator

gen = HybridArabicMathGenerator(seed=42)
batch = gen.generate_dataset_batch(count=2)

print("=== عينة من مخرجات المولد الرمزي الهجين المجهزة لتدريب Qwen2.5-Math ===\n")
for i, sample in enumerate(batch, 1):
    print(f"المسألة [{i}]: {sample['equation']}")
    print("المخرجات المهيكلة للنموذج:")
    print(sample['target'])
    print("-" * 50)
```

### 1.5 تجربة خدمة الـ API وتطبيق Flutter الإنتاجي
- **الهدف:** اختبار واجهة FastAPI للربط المباشر مع تطبيق الهاتف المحمول (Flutter Client).

```python
from khwarizmi.api.app import app
from khwarizmi.api.schemas import SolveRequest
from fastapi.testclient import TestClient

client = TestClient(app)
response = client.post("/api/v1/solve", json={"equation_latex": "3x - 9 = 15"})
data = response.json()

print(f"API Status Code: {response.status_code}")
print(f"Equation: {data['equation_latex']}")
print(f"Final Answer: {data['final_answer']}")
print(f"Computation Time: {data['computation_time_ms']} ms")
print(f"Total Verified Steps: {data['total_steps']}")

for step in data['steps']:
    print(f"  • [الخطوة {step['step_num']}] {step['rule_name_ar']}: {step['latex']}")
```
